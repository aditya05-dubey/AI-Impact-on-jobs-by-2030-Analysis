# Exploratory Data Analysis (EDA)

EDA is performed to understand the distribution of variables and identify patterns, relationships, and trends related to salary, skills, AI exposure, and automation risk.

## Analysis 1: Distribution of Automation Risk Categories

This analysis examines how workers are distributed across different automation risk levels (Low, Medium, and High).

```python
# Distribution of Automation Risk Categories

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='Risk_Category'
)

plt.title('Distribution of Automation Risk Categories')
plt.xlabel('Risk Category')
plt.ylabel('Number of Workers')

plt.show()
```

### Insight

The majority of workers fall into the Medium automation risk category, accounting for nearly half of the workforce. High-risk and low-risk groups are distributed almost equally.

This suggests that while a significant portion of workers may not face immediate replacement by AI, many jobs are likely to undergo transformation and require continuous upskilling and adaptation by 2030.

## Analysis 2: Education Level vs Salary

This analysis investigates whether educational qualifications influence salary levels across different job roles.

```python
# Education Level vs Salary

plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x='Education_Level',
    y='Average_Salary'
)

plt.title('Education Level vs Average Salary')
plt.xlabel('Education Level')
plt.ylabel('Average Salary')

plt.show()
```

### Insight

The salary distributions across all education levels appear very similar, with median salaries ranging between approximately $85,000 and $95,000.

Master's degree holders show a slightly higher median salary compared to other groups. However, the significant overlap between the salary distributions suggests that education level alone may not be the primary factor influencing salary in this dataset.

Other factors such as job role, experience, AI exposure, and skill levels may have a stronger impact on earnings.

## Analysis 3: Years of Experience vs Salary

This analysis explores the relationship between work experience and salary levels.

```python
# Experience vs Salary

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='Years_Experience',
    y='Average_Salary'
)

plt.title('Years of Experience vs Average Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Average Salary')

plt.show()
```

```python
df[['Years_Experience', 'Average_Salary']].corr()
```

### Insight

The correlation between Years of Experience and Average Salary is 0.017, indicating an extremely weak positive relationship.

The scatter plot shows that employees with both low and high experience levels can earn similar salaries. This suggests that experience alone is not a strong predictor of salary within this dataset.

Other factors such as job title, education level, AI exposure, technical growth potential, and skill proficiency may have a greater influence on compensation than years of experience.

## Analysis 4: AI Exposure vs Automation Probability

This analysis examines whether jobs with higher AI exposure are more likely to face automation by 2030.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='AI_Exposure_Index',
    y='Automation_Probability_2030'
)

plt.title('AI Exposure vs Automation Probability')
plt.xlabel('AI Exposure Index')
plt.ylabel('Automation Probability 2030')

plt.show()
```

```python
df[['AI_Exposure_Index','Automation_Probability_2030']].corr()
```

Insight

The scatter plot and correlation analysis reveal almost no relationship between AI Exposure Index and Automation Probability 2030 (correlation = 0.014).

This indicates that jobs with higher AI exposure are not necessarily more likely to be automated. AI may act as a productivity-enhancing tool in many occupations rather than a direct replacement for human workers.

Therefore, automation risk appears to depend on additional factors such as skill requirements, education level, and job-specific characteristics.

# Analysis 5: Education Level vs Average Salary

```python
# Education Level vs Average Salary

plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x='Education_Level',
    y='Average_Salary'
)

plt.title('Education Level vs Average Salary')

plt.show()
```

```python
df.groupby('Education_Level')['Average_Salary'].mean().sort_values(ascending=False)
```

### Insight

Education level has a slight impact on average salary in this dataset.

Employees with a Master's degree earn the highest average salary (~91,319), followed by PhD holders (~89,046). Workers with Bachelor's degrees and High School education show slightly lower average salaries.

However, the salary differences between education levels are relatively small, suggesting that education alone may not be a strong determinant of salary.

# Analysis 6: Distribution of Automation Risk Categories

This analysis examines the distribution of workers across Low, Medium, and High automation risk categories.

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='Risk_Category',
    order=['Low','Medium','High']
)

plt.title('Distribution of Automation Risk Categories')

plt.show()
```

### Insight

The majority of workers fall into the Medium automation risk category, accounting for approximately half of the dataset.

The Low and High risk categories contain a similar number of workers, indicating a balanced distribution between jobs that are relatively safe from automation and those that are highly vulnerable.

This suggests that most occupations are expected to experience some level of AI-driven change by 2030, but only a smaller proportion are expected to be either completely safe or highly exposed to automation.

# Analysis 7: Top 10 Jobs with Highest Automation Risk

This analysis identifies occupations that are most likely to be automated by 2030 based on their average automation probability.

```python
top_jobs = (
    df.groupby('Job_Title')['Automation_Probability_2030']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

top_jobs.plot(kind='bar')

plt.title('Top 10 Jobs Most Likely to be Automated by 2030')
plt.ylabel('Average Automation Probability')
plt.xlabel('Job Title')

plt.xticks(rotation=45)

plt.show()

top_jobs
```

### Insight

The analysis shows that Retail Workers, Security Guards, Construction Workers, Customer Support professionals, and Truck Drivers have the highest automation probabilities, with values above 0.80.

These occupations typically involve repetitive, predictable, and rule-based tasks, making them more susceptible to automation technologies and AI-driven systems.

In contrast, occupations such as Data Scientists, Marketing Managers, Chefs, and Software Engineers exhibit lower automation probabilities, suggesting that jobs requiring creativity, problem-solving, and advanced technical skills may remain less vulnerable to automation by 2030.

# Analysis 8: Top 10 Jobs with Highest AI Exposure

This analysis identifies occupations that are expected to interact most heavily with AI technologies in the future.

```python
top_ai_exposure = (
    df.groupby('Job_Title')['AI_Exposure_Index']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

top_ai_exposure.plot(kind='bar')

plt.title('Top 10 Jobs with Highest AI Exposure')
plt.ylabel('Average AI Exposure Index')
plt.xlabel('Job Title')

plt.xticks(rotation=45)

plt.show()

top_ai_exposure
```

### Insight

The AI Exposure Index is relatively similar across the top occupations, with values ranging between 0.50 and 0.53.

Chefs, Graphic Designers, Marketing Managers, and Nurses show the highest AI exposure, indicating that AI technologies are expected to increasingly support decision-making, creativity, and operational tasks in these professions.

However, the small differences between occupations suggest that AI adoption is expected across a wide range of industries rather than being concentrated in a few specific job roles.

# Analysis 9: Correlation Heatmap

This analysis examines the relationships among all numerical variables in the dataset to identify key factors associated with salary, automation risk, and AI exposure.

```python
plt.figure(figsize=(12,8))

corr_matrix = df.select_dtypes(include=['int64','float64']).corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

### Insight

The correlation heatmap reveals that most numerical variables have correlation values very close to zero.

Years of Experience, Average Salary, AI Exposure Index, Automation Probability, and Skill metrics show little to no linear relationship with one another.

The strongest correlations observed are still very weak (close to ±0.05), indicating that no single variable strongly influences another within this dataset.

This suggests that the dataset is highly balanced and may be synthetically generated, making it useful for exploratory analysis and visualization practice but less suitable for discovering strong real-world relationships.
